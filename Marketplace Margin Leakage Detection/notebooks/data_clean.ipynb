```python
import pandas as pd
import numpy as np
import os
import warnings

warnings.filterwarnings('ignore')
print(os.getcwd())
```

```python
# Define folder paths

RAW_PATH = r"D:\Data Analytics\Marketplace Margin Leakage Detection\datasets"
CLEAN_PATH = r"D:\Data Analytics\Marketplace Margin Leakage Detection\clean_datasets"

# Create cleaned folder if not exists

os.makedirs(CLEAN_PATH, exist_ok=True)


# Load all CSV files

datasets = {}

for file in os.listdir(RAW_PATH):

    if file.endswith('.csv'):

        file_path = os.path.join(RAW_PATH, file)

        df = pd.read_csv(file_path)

        datasets[file.replace('.csv', '')] = df

        print(f"{file} - Shape: {df.shape}")
```

```python
def data_profile(df, name):

    print(f"\n{'='*50}")
    print(f"DATASET: {name}")
    print(f"{'='*50}")

    print("Rows & Columns:", df.shape)

    print("\nColumn Information:")
    print(df.info())

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Records:")
    print(df.duplicated().sum())

    print("\nStatistical Summary:")
    display(df.describe().T)

for name, df in datasets.items():

    data_profile(df, name)
```

```python
def clean_column_names(df):

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(' ', '_', regex=False)
        .str.replace(r'[^a-zA-Z0-9_]', '', regex=True)
    )

    return df

def remove_duplicates(df):

    before = len(df)

    df = df.drop_duplicates()

    after = len(df)

    print(f"Duplicates Removed: {before - after}")

    return df
```

```python
def handle_missing_values(df):

    for col in df.columns:

        if df[col].dtype == 'object':

            df[col] = df[col].fillna('Unknown')

        elif pd.api.types.is_numeric_dtype(df[col]):

            df[col] = df[col].fillna(df[col].median())

    return df

def clean_text_columns(df):

    for col in df.select_dtypes(include='object').columns:

        df[col] = df[col].str.strip()

    return df
```

```python
def standardize_missing_values(df):

    df.replace(
        ['', ' ', 'NA', 'N/A', 'null', 'NULL', 'None', '?'],
        np.nan,
        inplace=True
    )

    return df
```

```python
# combine all function to check it 

def clean_dataset(df):

    df = df.copy()

    # 1. Standardize column names
    df = clean_column_names(df)

    # 2. Standardize missing values
    df = standardize_missing_values(df)

    # 3. Clean text columns
    df = clean_text_columns(df)

    # 4. Remove duplicates
    df = remove_duplicates(df)

    # 5. Handle missing values
    df = handle_missing_values(df)

    return df
```

```python
cleaned_datasets = {}

for name, df in datasets.items():

    print(f"\nCleaning Dataset: {name}")

    cleaned_df = clean_dataset(df)

    cleaned_datasets[name] = cleaned_df

    print("Cleaning Completed!")
```

```python
# save all dataset 
for name, df in cleaned_datasets.items():

    output_path = os.path.join(
        CLEAN_PATH,
        f"{name}_cleaned.csv"
    )

    df.to_csv(output_path, index=False)

    print(f"{name} Exported Successfully!")
```

```python
# generate a clening summury report

cleaning_report = []

for name in datasets:

    raw_df = datasets[name]
    clean_df = cleaned_datasets[name]

    cleaning_report.append({

        'Dataset': name,

        'Original_Rows': len(raw_df),

        'Cleaned_Rows': len(clean_df),

        'Original_Columns': len(raw_df.columns),

        'Cleaned_Columns': len(clean_df.columns),

        'Duplicates_Remaining': clean_df.duplicated().sum(),

        'Missing_Values_Remaining': clean_df.isnull().sum().sum()

    })


report_df = pd.DataFrame(cleaning_report)

display(report_df)
```
