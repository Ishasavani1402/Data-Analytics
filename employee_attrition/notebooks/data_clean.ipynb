```python
import pandas as pd
```

```python
df = pd.read_csv('D:\\Data Analytics\\employee_attrition\\dataset\\employee_attrition_hr_2026.csv')
df.head()
```

```python
df.info()
print("--"* 20)
print(f"null chheck : \n{df.isnull().sum()}")
print("--"* 20)
print(f"duplicate check : \n{df.duplicated().sum()}")
print("--"* 20)
print(f" data shape : {df.shape}")
```

```python
# remove white space from object 
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

```python
# check unique values
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    vals = df[col].unique()
    print(f"\n{col} ({len(vals)} unique): {sorted(vals)}")
```

```python
# Convert Yes/No to 1/0
df['attrition'] = df['attrition'].map({'Yes': 1, 'No': 0})

# Verify the conversion
print(df['attrition'].value_counts())
print(df['attrition'].dtype)
```

```python
df['gender'].value_counts()
```

**Removed 98 records where gender = "Prefer not to say" (1.96% of the dataset) due to low representation and limited analytical value.**

```python
df = df[df['gender'] != 'Prefer not to say']
```

```python
print('after remove prefere no to say :')
print(df['gender'].value_counts())
print("==========")
print(df.shape)
```

```python
df['job_level'].value_counts()
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[20, 29, 39, 49, 64],
    labels=[
        'Early Career (21-29)',
        'Mid Career (30-39)',
        'Senior Career (40-49)',
        'Pre-Retirement (50-64)'
    ]
)

job_level_map = {
    1: 'Entry Level',
    2: 'Associate',
    3: 'Mid Level',
    4: 'Senior Level',
    5: 'Management'
}

df['job_level'] =  df['job_level'].map(job_level_map)
```

```python
df['job_level'].value_counts()
```

```python
df.columns
```

```python
remove_cols = ['years_since_promotion' , 'num_prior_companies' , 'business_travel_days_per_year' , 'performance_rating' , 'last_review_score']

df = df.drop(columns = remove_cols)
```
