# python data cleaning of sales analysis 

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel(r'D:\Data Analytics\sales_analysis\dataset\Sales_Data.xlsx' , sheet_name='Sales Data')
df.head(5)
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ' , '_')
df.columns
```

```python
df.info()
print('=='*20)
print('null check')
print('=='*20)
print(df.isnull().sum())
print('=='*20)
print('duplicate check')
print('=='*20)
print(df.duplicated().sum())
print('=='*20)
print('data shape before cleaning')
print('=='*20)
print(df.shape)
```

```python
# sort missing 
missing = df.isnull().sum()
missing[missing > 0].sort_values(ascending=False)
```

* to check fully missing row

```python
df[df.isnull().all(axis=1)]
```

* remove this full miss row

```python
df = df.dropna(how='all').reset_index(drop=True)
print('data shape after remove full miss row')
df.shape
```

* to check which row exact duplicate

```python
# Display all rows involved in duplication
duplicate_rows = df[df.duplicated(keep=False)]
duplicate_rows
```

* drop duplicate rows

```python
# 1014 - 1 - 7
df = df.drop_duplicates().reset_index(drop=True)
print('after remove duplicate data shape')
df.shape
```

* handle missing customer name

```python
name_lookup = df.dropna(subset=['customer_name']).drop_duplicates('customer_id').set_index('customer_id')['customer_name']

df['customer_name'] = df['customer_name'].fillna(df['customer_id'].map(name_lookup))

df['customer_name'] = df['customer_name'].fillna('Unknown Customer')

df['customer_name'].isnull().sum()
```

* handle missing sale price

```python
mask = df['sales_price'].isnull()
df.loc[mask, 'sales_price'] = df.loc[mask, 'total_sales_amount'] / df.loc[mask, 'quantity'] 

df['sales_price'].isnull().sum()
```

* handle missing ship mode

```python
# For every order_id, find its first available ship_mode, then use that value to fill missing ship_mode values for the same order.
correct_ship_mode = (
    df.groupby('order_id')['ship_mode'].transform('first')
)

df['ship_mode'] = df['ship_mode'].fillna(correct_ship_mode)
```

* handle missing postal code

```python
state_zip_mode = df.dropna(subset=['postal_code']).groupby('state')['postal_code'].agg(lambda x: x.mode()[0])
df['postal_code'] = df['postal_code'].fillna(df['state'].map(state_zip_mode))

df['postal_code'].isnull().sum()
```

* handle missing product category

```python
cat_lookup = df.dropna(subset=['product_category']).drop_duplicates('sub-category').set_index('sub-category')['product_category']
df['product_category'] = df['product_category'].fillna(df['sub-category'].map(cat_lookup))
```

* handle missing order date

```python
# For every order_id, find its first available order_date, then use that value to fill missing order_date values for the same order.
correct_order_date= (
    df.groupby('order_id')['order_date'].transform('first')
)

df['order_date'] = df['order_date'].fillna(correct_order_date)

df['order_date'].isnull().sum()
```

```python
print('after handle null value :')
df.isnull().sum()
```

```python
df = df.rename(columns={'sub-category' :'sub_category'})
df.columns
```

* to check if negetive value exist in numeric_col

```python
numeric_col = df.select_dtypes(include=('float64' , 'int64'))
check_cols = numeric_col.drop(columns=['discount','shipping_days'])

negative_rows = df[(check_cols <= 0).any(axis=1)]
negative_rows

# here use < 0 because in discount can possible 0 .
```

```python
# remove white space from object 
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

```python
# check unique values
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    vals = df[col].unique()
    print(f"\n{col} ({len(vals)} unique): {sorted(vals)}")
```

* to check year column and order year consistency

```python
year_from_order_date = df['order_date'].dt.year

missmatch = df['year'] != year_from_order_date
print(f'year missmatch for year column and order date col : {missmatch.sum()}')

# Preview rows where Year disagrees with Order Date
df.loc[
    missmatch,
    ['order_id', 'year', 'order_date']
].assign(
    Correct_Year=year_from_order_date[missmatch]
).head(10)
```

```python
# Correct Year using the trusted Order Date column
df['year'] = df['order_date'].dt.year

# Confirm that all Year values now match Order Date
year_mismatch_after = (
    df['year'] != df['order_date'].dt.year
).sum()

print("Year Mismatches After Cleaning:", year_mismatch_after)
```

* here discount column all value are 0 that is complete wrong so need to calculate proper discount 

```python
df['discount'] = 1 - (df['total_sales_amount'] / (df['sales_price'] * df['quantity']))
df['discount'] = df['discount'].round(2) 
df['discount'].value_counts()
```

```python
# now check total sale amount = sale price * qty - discount check data consistency
calculated_total = (
    df['sales_price'] * df['quantity']
) - df['discount']

mismatch = ~np.isclose(
    df['total_sales_amount'],
    calculated_total,
    rtol=0,
    atol=0.01
)

inconsistent_rows = df[mismatch]

inconsistent_rows
print(f"Truly inconsistent rows (can't be explained by price x qty x discount): {mismatch.sum()}")
print(f"Rows where original 'discount' column was just wrong: {(df['discount'] != df['discount']).sum()}")
 
```

```python
# now check net profit = total sale amount - total purchase amount
#  check data consistency
calculated_profit = df['total_sales_amount'] - df['total_purchasing_amounts']

mismatch = ~np.isclose(
    df['net_profit'],
    calculated_profit,
    rtol=0,
    atol=0.01
)

inconsistent_rows = df[mismatch]

inconsistent_rows
```

* create a new column 

```python
df['shipping_days'] = (df['ship_date']-df['order_date']).dt.days
df['profit_margin_%'] = (df['net_profit'] / df['total_sales_amount']) * 100
```
